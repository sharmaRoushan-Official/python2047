### TypeCase

1. PascalCase: PascalCase is used to create class and written as first charcter is in capital and middle of character is capital
2. snake_case: snake_case is used to create varaible and method and wrritten as smaller character `_` used in middle.
3. camelCase: camelCase is used to create Method Name and written as small character and in the middle `capital` character

```python
numVar = 100 # camel
print(numVar)
```

```python
NumVar = 1200 # pascal
print(NumVar)
```

### Operators in Python

1. Arithematic Operator:- [+,-,*,/,//,%]
    1. Unary Operator: These operator work on single operand. Ex: i++, ++i,i--,--i
       * Pre Operator: ++i, --i
       * Post Operator:i++, i--
    2. Binary Operator: These Operator work on more than one Operands
        * 10+2, 45+89+78
2. Assignment Operator: [=]
   * x = 20 x is variable 20 is assigned value in x
  
3. Membership Operator: [in, not]
4. identity Operator: [is , is not]
5. Logical Operator: [`AND,OR,NOT`,NAND,NOR,XOR,XNOR]
   * Basic Gates: `AND,OR,NOT`
   * Universal Gates: `NAND,NOR`
