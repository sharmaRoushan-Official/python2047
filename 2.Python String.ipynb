### 2. String in Python

#### 1. What is a string in python?

A `String` in ptyhon is a sequence of Unicode chracters enclosed in quotes.

# hello
## hello 
###  hello
#### hello
##### hello
###### hello

```python
a 
```

```python
s1 = 'a' # single quote
s2 = "apple" # double quote
s3 = """hello world"""  ### triple quote # multi line comment
print(s1)
print(s2)
print(s3)
```

Notes:

> Strings are immutable(cannot be changed after creating)

> Indexed() : Ordered collection of data

> Support slicing, iteration, and many built-in methods

### String Inexing[Ordered collection]

Each chracters has a position

|-|P|y|t|h|o|n|
|---|---|---|---|---|---|---|
|Positive|0|1|2|3|4|5|
|Negative|-6|-5|-4|-3|-2|-1|

```python
x = "Python"

# Positive 
print(x[0])  # var[0]
print(x[1])
print(x[2])
print(x[3])
print(x[4])
print(x[5])
```

```python
# Negative Indexing

print(x[-1])
print(x[-2])
print(x[-3])
print(x[-4])
print(x[-5])
print(x[-6])
```

### Slicing of string

var[start:end]

```python
x = "Hello world"

print(x)
```

```python
y = x[0:5]
print(y)
```

```python
y = x[6:11]
print(y)
```

```python
x = "Hello world"

print(x[:])  # assign value by default
```

```python
x[:5]
```

```python
x[6:]
```

### Striding String

var[start:end:stride]

```python
x1 = "Hello world"

x1[:]  # slicing
```

```python
x1[0:11:2]
```

```python
x1[0:11:3]
```

```python
name = "LAlit"
print(name)
```

### 7. String Method

----

Case Conversion

```python
s = "hELLo WoRLd"
print(s)
```

```python
### upper case()

upr_case = s.upper()
print(upr_case)
```

```python
## lowerCase

lwr_case = s.lower()
print(lwr_case)
```

```python
### Capitalize

capitalzeChar = s.capitalize()
print(capitalzeChar)
```

```python
### title()

titleChar = s.title()
print(titleChar)
```

```python
# swapcase() 
print(s)
```

```python
swapCase = s.swapcase()
print(swapCase)
```
